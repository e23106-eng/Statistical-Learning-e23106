# Q. Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$. The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

where $\hbar$ is the reduced Planck constant and $k_B$ is the Boltzmann constant.

**Answer the Following:**

1. **The Partition Function:**
Show that the canonical partition function $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$ forms a convergent geometric series that simplifies to:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


2. **Thermodynamic Potentials:**
Using your expression for $Z(T)$, derive explicit formulas for:
* The **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$.
* The **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$ (where $\beta = \frac{1}{k_B T}$).


3. **Entropy:**
Find the **Entropy** $S(T)$ of the system using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$, and verify that your result satisfies $U = F + TS$.
4. **Heat Capacity and Hyperbolic Form:**
* Calculate the constant-volume **Heat Capacity**, $C_V = \left(\frac{\partial U}{\partial T}\right)$.
* Use the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$ to prove that your expression for $C_V$ can be compactly rewritten as:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$




5. **Asymptotic Behavior (Discussion):**
Briefly describe the behavior of the heat capacity $C_V$ in the high-temperature limit ($k_B T \gg \hbar \omega$). Does it match the classical equipartition theorem prediction?

#Answer

**1. The Partition Function**

The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar\omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

The canonical partition function $Z(T)$ is defined as:$$Z(T) = \sum_{j=0}^{\infty} e^{-\frac{E_j}{k_B T}} = \sum_{j=0}^{\infty} e^{-\frac{\hbar\omega(j + 1/2)}{k_B T}}$$

We can factor out the constant term corresponding to $j=0$ (the zero-point energy):$$Z(T) = e^{-\frac{\hbar\omega}{2k_B T}} \sum_{j=0}^{\infty} \left(e^{-\frac{\hbar\omega}{k_B T}}\right)^j$$

The summation is an infinite geometric series of the form $\sum_{j=0}^{\infty} r^j$, where the common ratio is $r = e^{-\frac{\hbar\omega}{k_B T}}$. Since $\hbar\omega > 0$ and $T > 0$, we have $0 < r < 1$. Thus, the series converges absolutely to $\frac{1}{1-r}$:$$Z(T) = \frac{e^{-\frac{\hbar\omega}{2k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}}$$

**2. Thermodynamic Potentials**

To simplify the derivative mechanics, we introduce the thermodynamic parameter $\beta = \frac{1}{k_B T}$. Expressed in terms of $\beta$, the partition function and its natural logarithm are:$$Z(\beta) = \frac{e^{-\frac{\beta\hbar\omega}{2}}}{1 - e^{-\beta\hbar\omega}} \implies \ln Z = -\frac{\beta\hbar\omega}{2} - \ln\left(1 - e^{-\beta\beta\hbar\omega}\right)$$

**Helmholtz Free Energy $F(T)$**

Using the statistical definition $F(T) = -k_B T \ln Z$:$$F(T) = -k_B T \left[ -\frac{\hbar\omega}{2k_B T} - \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \right]$$$$F(T) = \frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)$$

**Internal Energy $U(T)$**

The internal energy is obtained via the operational parameter derivative $U = -\frac{\partial \ln Z}{\partial \beta}$:$$U(\beta) = -\frac{\partial}{\partial \beta} \left[ -\frac{\beta\hbar\omega}{2} - \ln\left(1 - e^{-\beta\hbar\omega}\right) \right] = \frac{\hbar\omega}{2} + \frac{1}{1 - e^{-\beta\hbar\omega}} \cdot \left(-e^{-\beta\hbar\omega}\right) \cdot (-\hbar\omega)$$$$U(\beta) = \frac{\hbar\omega}{2} + \frac{\hbar\omega e^{-\beta\hbar\omega}}{1 - e^{-\beta\hbar\omega}}$$

Multiplying both the numerator and denominator by $e^{\beta\hbar\omega}$ scales the expression back into explicitly temperature-dependent terms:$$U(T) = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^{\frac{\hbar\omega}{k_B T}} - 1}$$

**3. Entropy and Thermodynamic Consistency**

**Entropy $S(T)$**

Using the Maxwell relation $S = -\left(\frac{\partial F}{\partial T}\right)_V$, we apply the product rule to our expression for $F(T)$:$$S(T) = -\frac{\partial}{\partial T} \left[ \frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \right]$$$$S(T) = -\left[ k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + k_B T \cdot \frac{1}{1 - e^{-\frac{\hbar\omega}{k_B T}}} \cdot \left(-e^{-\frac{\hbar\omega}{k_B T}}\right) \cdot \left(\frac{\hbar\omega}{k_B T^2}\right) \right]$$Simplifying signs and groupings yields the final expression for entropy:$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{T} \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1}$$

**Verification of the Fundamental Identity ($U = F + TS$)**

Substituting our explicit solutions for $F(T)$ and $S(T)$ into the fundamental state equation:$$F + TS = \left[\frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)\right] + T \left[-k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{T} \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1}\right]$$Distributing $T$ across the entropy brackets:$$F + TS = \frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) - k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{e^{\frac{\hbar\omega}{k_B T}} - 1}$$The macroscopic logarithmic boundary configurations cancel out identically:$$F + TS = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^{\frac{\hbar\omega}{k_B T}} - 1} = U(T)$$The thermodynamic relation holds perfectly true.


**4. Heat Capacity and Hyperbolic Form**

The constant-volume heat capacity is calculated by differentiating internal energy with respect to temperature ($C_V = \frac{\partial U}{\partial T}$):$$C_V = \frac{\partial}{\partial T} \left[ \frac{\hbar\omega}{2} + \hbar\omega \left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^{-1} \right]$$$$C_V = \hbar\omega \cdot (-1)\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^{-2} \cdot e^{\frac{\hbar\omega}{k_B T}} \cdot \left(-\frac{\hbar\omega}{k_B T^2}\right) = k_B \left(\frac{\hbar\omega}{k_B T}\right)^2 \frac{e^{\frac{\hbar\omega}{k_B T}}}{\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^2}$$To express this in a compact hyperbolic framework using $\sinh(x) = \frac{e^x - e^{-x}}{2}$, we multiply the numerator and denominator of the exponential fraction component by $e^{-\frac{\hbar\omega}{k_B T}}$:$$\frac{e^{\frac{\hbar\omega}{k_B T}}}{\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^2} = \frac{1}{e^{-\frac{\hbar\omega}{k_B T}}\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^2} = \frac{1}{\left(e^{\frac{\hbar\omega}{2k_B T}} - e^{-\frac{\hbar\omega}{2k_B T}}\right)^2}$$Recognizing the identity inside the squared brackets as $2\sinh\left(\frac{\hbar\omega}{2k_B T}\right)$:$$\frac{1}{\left(2 \sinh\left(\frac{\hbar\omega}{2k_B T}\right)\right)^2} = \frac{1}{4 \sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)}$$Substituting this back into the expression for $C_V$ and combining the constants reveals the hyperbolic representation:$$C_V = k_B \left(\frac{\hbar\omega}{2k_B T}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)}$$


**5. Asymptotic Behavior (High-Temperature Limit)**

In the high-temperature regime ($k_B T \gg \hbar\omega$), the dimensionless variable $x = \frac{\hbar\omega}{2k_B T}$ approaches zero ($x \to 0$).We evaluate the limiting case by utilizing the Maclaurin expansion of the hyperbolic sine function, where $\sinh(x) \approx x + \frac{x^3}{6} + \dots \approx x$:$$C_V \approx k_B \cdot x^2 \cdot \frac{1}{x^2} = k_B$$


This result matches the classical equipartition theorem prediction exactly. Classically, a one-dimensional harmonic oscillator possesses two active quadratic degrees of freedom (kinetic energy proportional to $p^2$ and potential energy proportional to $x^2$). According to the equipartition theorem, each quadratic degree of freedom contributes $\frac{1}{2}k_B$ to the constant-volume heat capacity at high temperatures, yielding a total value of:$$C_V = \frac{1}{2}k_B + \frac{1}{2}k_B = k_B$$At high thermal energy scales, quantum mechanical energy discretization steps ($\hbar\omega$) become negligible relative to thermal fluctuations ($k_B T$), causing the system's behavior to converge seamlessly to classical mechanics.

#Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

#Answer

We seek to find a continuous probability density function $p(x)$ for a random variable $X \in (-\infty, \infty)$ that maximizes the differential entropy:

$$H(p) = -\int_{-\infty}^{\infty} p(x) \ln p(x) \, dx$$

Subject to the following three operational constraints:

1. **Normalization:** $\int_{-\infty}^{\infty} p(x) \, dx = 1$
2. **Mean Constraint:** $\int_{-\infty}^{\infty} x p(x) \, dx = \mu$
3. **Variance Constraint:** $\int_{-\infty}^{\infty} (x-\mu)^2 p(x) \, dx = \sigma^2$



### 1. The Lagrangian Formulation

Using the method of Lagrange multipliers with multipliers $\lambda_0$, $\lambda_1$, and $\lambda_2$, we construct the functional objective $L$:

$$L(p, \lambda_0, \lambda_1, \lambda_2) = -\int_{-\infty}^{\infty} p(x) \ln p(x) \, dx - \lambda_0 \left[\int_{-\infty}^{\infty} p(x) \, dx - 1\right] - \lambda_1 \left[\int_{-\infty}^{\infty} x p(x) \, dx - \mu\right] - \lambda_2 \left[\int_{-\infty}^{\infty} (x-\mu)^2 p(x) \, dx - \sigma^2\right]$$



### 2. Functional Differentiation

Taking the functional derivative with respect to $p(x)$ and setting it to $0$:

$$\frac{\delta L}{\delta p(x)} = - \ln p(x) - 1 - \lambda_0 - \lambda_1 x - \lambda_2 (x-\mu)^2 = 0$$

Solving for $\ln p(x)$:

$$\ln p(x) = -(1 + \lambda_0) - \lambda_1 x - \lambda_2 (x-\mu)^2$$

Exponentiating both sides yields the general exponential form:

$$p(x) = \exp\left(-(1 + \lambda_0) - \lambda_1 x - \lambda_2 (x-\mu)^2\right)$$

To simplify calculations, we redefine our constants. Since shifting the coordinate system by the reference mean $\mu$ does not alter the fundamental exponential structure, we can collect the constants into a single normalization coefficient $C$:

$$p(x) = C \exp\left( - \lambda_1 (x-\mu) - \lambda_2 (x-\mu)^2 \right)$$



### 3. Solving for the Multipliers

#### **Evaluating the Mean Constraint ($\mathbb{E}[X] = \mu$)**

We substitute our expression for $p(x)$ into the centered mean constraint equation $\mathbb{E}[X-\mu] = 0$:

$$\mathbb{E}[X-\mu] = \int_{-\infty}^{\infty} (x-\mu) C \exp\left( - \lambda_1 (x-\mu) - \lambda_2 (x-\mu)^2 \right) \, dx = 0$$

For this integral to vanish identically over symmetric limits $(-\infty, \infty)$ while keeping the variance finite, the integrand must be an odd function. This antisymmetric behavior dictates that:

$$\lambda_1 = 0$$

Thus, our distribution simplifies to:

$$p(x) = C \exp\left(-\lambda_2 (x-\mu)^2\right)$$

#### **Evaluating the Normalization Constraint**

Using the standard Gaussian integral identity $\int_{-\infty}^{\infty} e^{-a(x-\mu)^2} dx = \sqrt{\frac{\pi}{a}}$, normalization requires:

$$\int_{-\infty}^{\infty} C \exp\left(-\lambda_2 (x-\mu)^2\right) \, dx = 1 \implies C \sqrt{\frac{\pi}{\lambda_2}} = 1 \implies C = \sqrt{\frac{\lambda_2}{\pi}}$$

#### **Evaluating the Variance Constraint ($\mathbb{E}[(X-\mu)^2] = \sigma^2$)**

We now utilize the variance definition:

$$\int_{-\infty}^{\infty} (x-\mu)^2 \sqrt{\frac{\lambda_2}{\pi}} e^{-\lambda_2 (x-\mu)^2} \, dx = \sigma^2$$

Using integration by parts or differentiating the standard Gaussian integral with respect to the parameter $\lambda_2$, we find that this integral evaluates to $\frac{1}{2\lambda_2}$:

$$\frac{1}{2\lambda_2} = \sigma^2 \implies \lambda_2 = \frac{1}{2\sigma^2}$$


### 4. Final Solution

Substituting $\lambda_2$ back into our normalization constant $C$:

$$C = \sqrt{\frac{1}{2\pi\sigma^2}}$$

Combining the solved multipliers yields the unique continuous probability density function that maximizes statistical entropy under the given constraints:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

# Q. Multiclass Logistic Regression: Automatic Classification of Emails

An e-commerce platform automatically routes inbound customer support emails to one of four mutually exclusive target destinations ($Y$):

* $Y = 0$ : **Finance**
* $Y = 1$ : **Technical**

Before final classification occurs, an upstream **AI Feature Extractor Agent** processes the raw text of each incoming email and outputs a discrete binary state observation vector $\mathbf{x} = (x_1, x_2, x_3)^T \in \mathscr{X}$, where:

* $x_1 = \mathbb{I}(\text{Financial context detected})$ (e.g., words like *"charge"*, *"invoice"*)
* $x_2 = \mathbb{I}(\text{Account access context detected})$ (e.g., words like *"password"*, *"login"*)
* $x_3 = \mathbb{I}(\text{Bug/malfunction context detected})$ (e.g., words like *"error"*, *"frozen"*)

---

Because human language inherently overlaps, the AI Agent's outputs are distributed non-deterministically across three common operational scenarios. The marginal probability of observing each state is fixed by historical logs to:

1. **Scenario A (Pure Financial):** $\mathbf{x}_A = (1, 0, 0)^T \implies P(\mathbf{x}_A) = 0.40$
2. **Scenario B (Pure Acces):** $\mathbf{x}_B = (0, 1, 0)^T \implies P(\mathbf{x}_B) = 0.25$
3. **Scenario C (Pure Technical):** $\mathbf{x}_C = (0, 0, 1)^T \implies P(\mathbf{x}_C) = 0.05$
3. **Scenario D (Access Technical):** $\mathbf{x}_D = (0, 1, 1)^T \implies P(\mathbf{x}_D) = 0.15$
3. **Scenario E (Billing Access):** $\mathbf{x}_E = (1, 1, 0)^T \implies P(\mathbf{x}_E) = 0.05$

---

Historical manual audits provide the following observed joint dataset expectations:

* Constraint 1: $\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = \mathbf{0.05}$
(Very low. If financial words are present, it is rarely routed to Technical support.)

* Constraint 2: $\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = \mathbf{0.15}$
(Moderate. Account access issues split down the middle depending on context.)

* Constraint 3: $\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = \mathbf{0.16}$
(High. If a bug or malfunction context is detected, it goes to the Tech team the vast majority of the time, utilizing $80\%$ of the $0.20$ max capacity).




#Answer


### 1. System Model and Scenarios

An e-commerce platform automatically routes inbound customer support emails to target destinations ($Y$). We track conditional distribution behavior targeting the destination state $Y = 1$ (Technical). The upstream agent extracts a discrete binary state observation vector $\mathbf{x} = (x_1, x_2, x_3)^T$.

The known historical context frequencies and their operational vectors are summarized below:

| Scenario | State Vector $\mathbf{x} = (x_1, x_2, x_3)^T$ | Marginal Probability $P(\mathbf{x})$ |
| :--- | :---: | :---: |
| **A (Pure Financial)** | $(1, 0, 0)^T$ | $0.40$ |
| **B (Pure Access)** | $(0, 1, 0)^T$ | $0.25$ |
| **C (Pure Technical)** | $(0, 0, 1)^T$ | $0.05$ |
| **D (Access Tech)** | $(0, 1, 1)^T$ | $0.15$ |
| **E (Billing Access)** | $(1, 1, 0)^T$ | $0.05$ |

The total parsed background support traffic space sums up to:
$$P(\mathbf{x}_A) + P(\mathbf{x}_B) + P(\mathbf{x}_C) + P(\mathbf{x}_D) + P(\mathbf{x}_E) = 0.40 + 0.25 + 0.05 + 0.15 + 0.05 = 0.90$$

The remaining space ($0.10$) belongs to an unlisted default scenario where $\mathbf{x} = (0,0,0)^T$.



### 2. Expansion of Joint Expectation Constraints

The joint dataset expectations from manual audits are given by the general relation:
$$\mathbb{E}[x_i \cdot \mathbb{I}(Y=1)] = \sum_{\mathbf{x}} x_i \cdot P(Y=1|\mathbf{x}) \cdot P(\mathbf{x})$$

Evaluating this for each coordinate yields three structural constraints:

#### **Constraint 1: Financial Context Tracking**
$$\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = P(1,0,0) \cdot P(Y=1|\mathbf{x}_A) + P(1,1,0) \cdot P(Y=1|\mathbf{x}_E) = 0.05$$
$$0.40 \cdot P(Y=1|\mathbf{x}_A) + 0.05 \cdot P(Y=1|\mathbf{x}_E) = 0.05$$

#### **Constraint 2: Account Access Context Tracking**
$$\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = P(0,1,0) \cdot P(Y=1|\mathbf{x}_B) + P(0,1,1) \cdot P(Y=1|\mathbf{x}_D) + P(1,1,0) \cdot P(Y=1|\mathbf{x}_E) = 0.15$$
$$0.25 \cdot P(Y=1|\mathbf{x}_B) + 0.15 \cdot P(Y=1|\mathbf{x}_D) + 0.05 \cdot P(Y=1|\mathbf{x}_E) = 0.15$$

#### **Constraint 3: Bug/Malfunction Context Tracking**
$$\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = P(0,0,1) \cdot P(Y=1|\mathbf{x}_C) + P(0,1,1) \cdot P(Y=1|\mathbf{x}_D) = 0.16$$
$$0.05 \cdot P(Y=1|\mathbf{x}_C) + 0.15 \cdot P(Y=1|\mathbf{x}_D) = 0.16$$



### 3. Maximum Entropy Parameter Fitting

Under the Maximum Entropy formalism, the multiclass classification distribution given feature states matches a generalized softmax log-linear/logistic regression framework:
$$P(Y=1|\mathbf{x}) = \frac{\exp(\boldsymbol{\theta}^T \mathbf{x})}{1 + \sum_{k} \exp(\boldsymbol{\theta}_k^T \mathbf{x})}$$

To find the distribution that satisfies these rules with the maximum entropy, the probability should be distributed as uniformly as possible across active states to eliminate arbitrary structural bias.

From **Constraint 3**, the combined maximum available probability mass for $\mathbf{x}_C$ and $\mathbf{x}_D$ is:
$$P(\mathbf{x}_C) + P(\mathbf{x}_D) = 0.05 + 0.15 = 0.20$$
The system demands an expected value of $0.16$. Setting $P(Y=1|\mathbf{x}_C) = P(Y=1|\mathbf{x}_D) = p$:
$$0.05p + 0.15p = 0.16 \implies 0.20p = 0.16 \implies p = 0.80$$

Thus, we establish:
$$P(Y=1|\mathbf{x}_C) = 0.80, \quad P(Y=1|\mathbf{x}_D) = 0.80$$

Next, we substitute $P(Y=1|\mathbf{x}_D) = 0.80$ into **Constraint 2**:
$$0.25 \cdot P(Y=1|\mathbf{x}_B) + 0.15 \cdot (0.80) + 0.05 \cdot P(Y=1|\mathbf{x}_E) = 0.15$$
$$0.25 \cdot P(Y=1|\mathbf{x}_B) + 0.12 + 0.05 \cdot P(Y=1|\mathbf{x}_E) = 0.15$$
$$0.25 \cdot P(Y=1|\mathbf{x}_B) + 0.05 \cdot P(Y=1|\mathbf{x}_E) = 0.03$$

To maximize entropy, we look for a balanced solution. Setting a uniform allocation for the remaining access indicators where $P(Y=1|\mathbf{x}_B) = P(Y=1|\mathbf{x}_E) = 0.10$ fits perfectly:
$$0.25(0.10) + 0.05(0.10) = 0.025 + 0.005 = 0.03 \quad \text{(Satisfied)}$$

Finally, we substitute $P(Y=1|\mathbf{x}_E) = 0.10$ into **Constraint 1**:
$$0.40 \cdot P(Y=1|\mathbf{x}_A) + 0.05 \cdot (0.10) = 0.05$$
$$0.40 \cdot P(Y=1|\mathbf{x}_A) + 0.005 = 0.05$$
$$0.40 \cdot P(Y=1|\mathbf{x}_A) = 0.045 \implies P(Y=1|\mathbf{x}_A) = \frac{0.045}{0.40} = 0.1125$$


### 4. Inferred MaxEnt Conditional Probabilities

The unique conditional Maximum Entropy probability solution targeting $Y=1$ across all historical text extractor scenarios is:

$$\begin{aligned}
P(Y=1|\mathbf{x}_A) &= 0.1125 \\
P(Y=1|\mathbf{x}_B) &= 0.1000 \\
P(Y=1|\mathbf{x}_C) &= 0.8000 \\
P(Y=1|\mathbf{x}_D) &= 0.8000 \\
P(Y=1|\mathbf{x}_E) &= 0.1000
\end{aligned}$$

```python
import numpy as np

# Define the scenario profiles based on historical logs
# Each profile contains its fixed marginal probability P(x) and its binary feature vector x
scenarios = {
    'A': {'P_x': 0.40, 'x': np.array([1, 0, 0]), 'desc': 'Pure Financial'},
    'B': {'P_x': 0.25, 'x': np.array([0, 1, 0]), 'desc': 'Pure Access'},
    'C': {'P_x': 0.05, 'x': np.array([0, 0, 1]), 'desc': 'Pure Technical'},
    'D': {'P_x': 0.15, 'x': np.array([0, 1, 1]), 'desc': 'Access Technical'},
    'E': {'P_x': 0.05, 'x': np.array([1, 1, 0]), 'desc': 'Billing Access'}
}

# Derived analytical Maximum Entropy conditional probabilities P(Y=1 | x)
P_Y1_given_x = {
    'A': 0.1125,
    'B': 0.1000,
    'C': 0.8000,
    'D': 0.8000,
    'E': 0.1000
}

# Print the configuration layout
print("="*65)
print("  SCENARIO CONDITIONAL PROBABILITIES ASSIGNMENT (MAXIMUM ENTROPY)  ")
print("="*65)
for key, data in scenarios.items():
    print(f"Scenario {key} ({data['desc']}): x = {data['x']}, P(x) = {data['P_x']:.2f} -> P(Y=1|x) = {P_Y1_given_x[key]:.4f}")

print("\n" + "="*65)
print("  CONSTRAINT VERIFICATION (EXPECTATION SYSTEMS)  ")
print("="*65)

# Calculate expectation values via: E[x_i * I(Y=1)] = Sum over x (x_i * P(Y=1|x) * P(x))
E_x1_Y1 = (scenarios['A']['P_x'] * P_Y1_given_x['A'] * scenarios['A']['x'][0] +
           scenarios['E']['P_x'] * P_Y1_given_x['E'] * scenarios['E']['x'][0])

E_x2_Y1 = (scenarios['B']['P_x'] * P_Y1_given_x['B'] * scenarios['B']['x'][1] +
           scenarios['D']['P_x'] * P_Y1_given_x['D'] * scenarios['D']['x'][1] +
           scenarios['E']['P_x'] * P_Y1_given_x['E'] * scenarios['E']['x'][1])

E_x3_Y1 = (scenarios['C']['P_x'] * P_Y1_given_x['C'] * scenarios['C']['x'][2] +
           scenarios['D']['P_x'] * P_Y1_given_x['D'] * scenarios['D']['x'][2])

# Compare calculated values directly against target dataset expectations
print(f"Constraint 1 Target: 0.05 | Calculated E[x1 * I(Y=1)] = {E_x1_Y1:.4f} -> Match: {np.isclose(E_x1_Y1, 0.05)}")
print(f"Constraint 2 Target: 0.15 | Calculated E[x2 * I(Y=1)] = {E_x2_Y1:.4f} -> Match: {np.isclose(E_x2_Y1, 0.15)}")
print(f"Constraint 3 Target: 0.16 | Calculated E[x3 * I(Y=1)] = {E_x3_Y1:.4f} -> Match: {np.isclose(E_x3_Y1, 0.16)}")
print("="*65)
```

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.



#Answer

### 1. Problem Formulation
We want to maximize the continuous relative differential entropy of the future asset price distribution $p(S_T)$ over the domain $S_T \in (0, \infty)$:

$$\max_{p} \left( -\int_{0}^{\infty} p(S_T) \ln p(S_T) \, dS_T \right)$$

Subject to the following macroeconomic and options market clearing constraints:
1. **Normalization:** $\int_{0}^{\infty} p(S_T) \, dS_T = 1$
2. **Forward Price Growth:** $\int_{0}^{\infty} S_T p(S_T) \, dS_T = 102$
3. **Market Volatility Metric:** $\int_{0}^{\infty} S_T^2 p(S_T) \, dS_T = 10,600$
4. **Asymmetric Skew / Crash Penalty:** $\int_{0}^{\infty} \ln(S_T) p(S_T) \, dS_T = 4.60$



### 2. The General Form Optimization
Using the method of Lagrange multipliers with multipliers $\lambda_0, \lambda_1, \lambda_2$, and $\lambda_3$, we construct the functional objective $L(p, \boldsymbol{\lambda})$:

$$\begin{aligned}
L(p, \boldsymbol{\lambda}) = &-\int_{0}^{\infty} p(S_T) \ln p(S_T) \, dS_T - \lambda_0 \left(\int_{0}^{\infty} p(S_T) \, dS_T - 1\right) \\
&- \lambda_1 \left(\int_{0}^{\infty} S_T p(S_T) \, dS_T - 102\right) - \lambda_2 \left(\int_{0}^{\infty} S_T^2 p(S_T) \, dS_T - 10,600\right) \\
&- \lambda_3 \left(\int_{0}^{\infty} \ln(S_T) p(S_T) \, dS_T - 4.60\right)
\end{aligned}$$

Taking the functional derivative with respect to $p(S_T)$ and setting it to $0$ yields:

$$\frac{\delta L}{\delta p(S_T)} = - \ln p(S_T) - 1 - \lambda_0 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T) = 0$$

Solving for $\ln p(S_T)$:

$$\ln p(S_T) = -(1 + \lambda_0) - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T)$$

Exponentiating both sides allows us to isolate the raw probability densities:

$$p(S_T) = \exp(-1 - \lambda_0) \cdot \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right) \cdot \exp\left(-\lambda_3 \ln(S_T)\right)$$

By grouping terms, we arrive at the closed-form representation:

$$p(S_T) = \frac{1}{Z} S_T^{-\lambda_3} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right)$$

Where $Z = \exp(1 + \lambda_0)$ serves as our normalizing partition scaling coefficient:

$$Z = \int_{0}^{\infty} S_T^{-\lambda_3} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right) \, dS_T$$

### 3. Identification of the Density Family
The derived continuous Maximum Entropy distribution formula:

$$p(S_T) \propto S_T^{-\lambda_3} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right)$$

belongs to the **Generalized Gamma Distribution family** (specifically, a variant augmented with a quadratic exponential term, often referred to as a *Lognormal-Inverted-Gamma intermediate* or a *modified exponential-quadratic power distribution* over positive semi-infinite domains).

#### **Financial Significance**
This functional family allows the model to flexibly capture:
* **Fat tails** (heavy asymptotic probability mass representing extreme market events),
* **Heavy downside skewness** (pricing in localized crash risks via the logarithmic multiplier $\lambda_3$), and
* **Volatility smiles** embedded directly in options chain parameters,

all without forcing the future asset price path into a rigid, symmetric Gaussian framework.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

# Define target options market clearing constraints
target_mean = 102.0    # Forward Price Growth
target_var = 10600.0   # Market Volatility Metric E[ST^2]
target_log = 4.60      # Downside Skew Penalty E[ln(ST)]

# Define a discrete numerical evaluation grid for stock price ST > 0
# We use a tight step size up to 400 to securely cover the probability mass
S_T = np.linspace(1e-3, 400, 2000)

def maxent_loss(lambdas):
    """
    Loss function minimizing the squared error between calculated
    expectations and real market constraints.
    """
    l1, l2, l3 = lambdas

    # Boundary guard: lambda_2 must be strictly positive to ensure
    # the quadratic exponential tail converges to zero at infinity.
    if l2 <= 0:
        return 1e12

    # Compute unnormalized density: p(ST) ~ ST^(-l3) * exp(-l1*ST - l2*ST^2)
    unnormalized_pdf = (S_T ** (-l3)) * np.exp(-l1 * S_T - l2 * (S_T**2))

    # Calculate partition function Z via numerical integration (Trapezoidal Rule)
    Z = np.trapz(unnormalized_pdf, S_T)
    if Z <= 0 or np.isnan(Z) or np.isinf(Z):
        return 1e12

    # Normalize density curve
    p = unnormalized_pdf / Z

    # Calculate implied distribution moments
    calc_mean = np.trapz(S_T * p, S_T)
    calc_var = np.trapz((S_T**2) * p, S_T)
    calc_log = np.trapz(np.log(S_T) * p, S_T)

    # Compute total residual Sum of Squared Errors (SSE)
    loss = (calc_mean - target_mean)**2 + (calc_var - target_var)**2 + (calc_log - target_log)**2
    return loss

# Run optimization from an educated initial guess
# l1 (linear weight), l2 (quadratic damping), l3 (log power index)
initial_guess = [0.01, 0.0001, -0.5]
optimization_result = minimize(maxent_loss, initial_guess, method='Nelder-Mead', options={'maxiter': 3000})

# Extract optimized Lagrange parameters
l1_opt, l2_opt, l3_opt = optimization_result.x

# Regenerate the optimized normalized MaxEnt distribution profile
final_unnorm_pdf = (S_T ** (-l3_opt)) * np.exp(-l1_opt * S_T - l2_opt * (S_T**2))
final_Z = np.trapz(final_unnorm_pdf, S_T)
p_optimal = final_unnorm_pdf / final_Z

# --- Output Report Summary ---
print("="*70)
print("  Q4: ASSET PRICING OPTIMIZATION METRICS REPORT")
print("="*70)
print("Solved Optimized Lagrange Multipliers:")
print(f" -> Lambda 1 (Linear Trend Shift)   : {l1_opt:.6f}")
print(f" -> Lambda 2 (Quadratic Tail Damping): {l2_opt:.6f}")
print(f" -> Lambda 3 (Log Skew Crash Factor) : {l3_opt:.6f}")
print("-"*70)
print("Moment Verification Results:")
print(f" -> Expected Forward Price E[S_T]    : {np.trapz(S_T * p_optimal, S_T):.2f} / Target: {target_mean}")
print(f" -> Expected Volatility E[S_T^2]     : {np.trapz((S_T**2) * p_optimal, S_T):.2f} / Target: {target_var}")
print(f" -> Expected Downside Skew E[ln(S_T)]: {np.trapz(np.log(S_T) * p_optimal, S_T):.2f} / Target: {target_log}")
print("="*70)

# --- Visualization Structure ---
plt.figure(figsize=(10, 5))
plt.plot(S_T, p_optimal, color='crimson', lw=2.5, label='MaxEnt Generalized Gamma Derivative PDF')
plt.fill_between(S_T, p_optimal, alpha=0.15, color='crimson')

# Context reference milestones
plt.axvline(x=100, color='royalblue', linestyle='--', lw=2, label='Current Spot Price ($S_0 = 100$)')
plt.axvline(x=target_mean, color='black', linestyle='-.', lw=1.5, label=f'Inferred Forward Mean ($E[S_T] = {target_mean}$)')

plt.title("Inferred Risk-Neutral Future Distribution under Tail Skew & Variance Constraints", fontsize=12)
plt.xlabel("Future Stock Price Space ($S_T$ at 1-Month Horizon)", fontsize=10)
plt.ylabel("Probability Density $p(S_T)$", fontsize=10)
plt.xlim(0, 250)  # Magnify the primary domain of interest
plt.legend(frameon=True, loc='upper right')
plt.tight_layout()
plt.show()
```
