**1. Finite-Sample Bias Derivation of $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$**

Let $\mathbf{X}_1, \mathbf{X}_2, \dots, \mathbf{X}_n \in \mathbb{R}^m$ be independent and identically distributed ($\text{i.i.d.}$) random vectors with mean vector $\boldsymbol{\mu}$ and population covariance matrix $\boldsymbol{\Sigma}$. The raw Maximum Likelihood Estimator ($\text{MLE}$) for the covariance matrix is defined as:$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

where $\widehat{\boldsymbol{\mu}}_n = \frac{1}{n}\sum_{i=1}^n \mathbf{X}_i$ is the empirical sample mean vector.To evaluate the expected value $\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}]$, we express the deviation of each observation from the sample mean by introducing the true latent population mean vector $\boldsymbol{\mu}$:$$\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n = (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$$Substituting this identity into the definition of $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ yields:$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right]^T$$Expanding the outer product inside the summation:$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n \left[ (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - (\mathbf{X}_i - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T + (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T \right]$$Distributing the summation operator $\sum_{i=1}^n$ across each individual term:$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - \frac{1}{n} \left( \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) \right) (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - \frac{1}{n} (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \left( \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) \right)^T + \frac{1}{n} \cdot n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$Note that by algebra, $\sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) = n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$. Substituting this simplification into the expression:$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - \frac{n}{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - \frac{n}{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T + (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$Combining the trailing identical outer-product matrices simplifies the equation down to:$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$Taking the mathematical expectation of both sides, and leveraging the linearity property of the expectation operator $\mathbb{E}[\cdot]$:$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} \sum_{i=1}^n \mathbb{E}\left[ (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T \right] - \mathbb{E}\left[ (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T \right]$$By definition of the population parameters:$\mathbb{E}\left[ (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T \right] = \boldsymbol{\Sigma}$ for all $i$.$\mathbb{E}\left[ (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T \right] = \text{Var}(\widehat{\boldsymbol{\mu}}_n) = \frac{1}{n}\boldsymbol{\Sigma}$.Substituting these known variance expressions back into the expectation equation:$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} \sum_{i=1}^n \boldsymbol{\Sigma} - \frac{1}{n}\boldsymbol{\Sigma} = \frac{1}{n} (n\boldsymbol{\Sigma}) - \frac{1}{n}\boldsymbol{\Sigma} = \boldsymbol{\Sigma} - \frac{1}{n}\boldsymbol{\Sigma} = \left(1 - \frac{1}{n}\right)\boldsymbol{\Sigma} = \frac{n-1}{n}\boldsymbol{\Sigma}$$

This proves that $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ is an inherently biased estimator in finite samples, systematically underestimating the true population covariance structure by a scale factor of $\frac{n-1}{n}$.

 **Derivation of the Unique, Unbiased Estimator $\mathbf{S}$ via Bessel's Correction**

To eliminate this down-scaling bias, we isolate the true population covariance matrix $\boldsymbol{\Sigma}$ from the derived expected value identity. We multiply both sides of our expectation equation by the reciprocal scaling factor $\frac{n}{n-1}$:

$$\frac{n}{n-1} \mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \boldsymbol{\Sigma} \implies \mathbb{E}\left[ \frac{n}{n-1} \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} \right] = \boldsymbol{\Sigma}$$

By expanding $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ within this expression, we arrive at the definition of our sample covariance matrix $\mathbf{S}$:$$\mathbf{S} = \frac{n}{n-1} \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{n}{n-1} \left( \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right) = \frac{1}{n-1} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

Thus, applying Bessel's correction factor ($n-1$ in the denominator) guarantees that:$$\mathbb{E}[\mathbf{S}] = \boldsymbol{\Sigma}$$This confirms that $\mathbf{S}$ is a uniquely unbiased estimator for the latent population covariance.

**2. Decision Space Dynamics & Error Control in Continuous Diagnostic Layers**

Type I Error ($\alpha$): The probability of committing a false alarm. In Structural Health Monitoring ($\text{SHM}$), this occurs when the test statistic exceeds the critical boundary threshold, causing the diagnostic layer to reject $H_0$ and flag an asset as damaged or compromised, when in reality it remains completely healthy.

$$\alpha = P(\text{Reject } H_0 \mid H_0 \text{ is True})$$

Type II Error ($\beta$): The probability of missing a true structural failure. In $\text{SHM}$, this occurs when an asset experiences structural damage (meaning $H_0$ is false and a alternative hypothesis $H_1$ is true), but the calculated test statistic fails to clear the critical threshold, leaving the system marked as nominal.

$$\beta = P(\text{Fail to Reject } H_0 \mid H_0 \text{ is False})$$

**Mathematical Consequences of Setting an Ultra-Conservative Threshold ($\alpha = 0.0001$)**

When an engineer sets an ultra-conservative threshold to minimize false alarms ($\alpha \to 0$), they must move the critical decision boundary out into the extreme tails of the null distribution.Because the statistical power of a test is defined as $1 - \beta$, this change creates a severe trade-off:
$$\text{Power} = 1 - \beta = \int_{\mathcal{R}} f(\mathbf{x} \mid H_1) \, d\mathbf{x}$$
where $\mathcal{R}$ is the rejection region dictated by $\alpha$. Shrinking $\alpha$ reduces the size of the rejection region $\mathcal{R}$. As a result, the integration area under the alternative hypothesis PDF $f(\mathbf{x} \mid H_1)$ shrinks as well. This causes a dramatic drop in statistical power ($1-\beta$), which increases the risk of a missed detection ($\beta$). The diagnostic layer becomes less sensitive, meaning it will likely miss early-stage structural changes or slow damage propagation until the anomalies grow large enough to force the statistic past the distant threshold.

**Geometric Volume Modifiers of the "Healthy Operation" Confidence Ellipsoid**

Under a Multivariate Gaussian null distribution assumption, the boundary of the healthy operating zone forms a confidence hyper-ellipsoid centered at $\widehat{\boldsymbol{\mu}}_n$. This boundary is defined by the Mahalanobis distance threshold $c$:
$$\mathcal{E}(\alpha) = \left\{ \mathbf{x} \in \mathbb{R}^m \mid (\mathbf{x} - \widehat{\boldsymbol{\mu}}_n)^T \mathbf{S}^{-1} (\mathbf{x} - \widehat{\boldsymbol{\mu}}_n) \le c^2(\alpha) \right\}$$

The critical threshold limit $c^2(\alpha)$ is determined by the inverse CDF of a chi-squared distribution, $\chi^2_m(1-\alpha)$. The geometric volume $V$ of this multi-dimensional confidence ellipsoid is directly proportional to the threshold radius raised to the power of the feature dimension $m$:
$$V(\mathcal{E}(\alpha)) = \frac{\pi^{m/2}}{\Gamma\left(\frac{m}{2} + 1\right)} \det(\mathbf{S})^{1/2} \left[ \chi^2_m(1 - \alpha) \right]^{m/2}$$

When $\alpha$ drops from $0.05$ to $0.0001$, the percentile target $1-\alpha$ climbs from $0.95$ to $0.9999$. This causes the critical threshold value $\chi^2_m(1-\alpha)$ to swell. Geometrically, this inflates the volume of the healthy confidence ellipsoid. The boundary expands outward, swallowing data points that would normally indicate early structural wear. This visually demonstrates why the system becomes less capable of detecting subtle anomalies.

### **Answer**

**1. Formal Statement of Slutsky's Theorem**

Let $\mathbf{Y}_n$ and $\mathbf{Z}_n$ be sequences of random vectors/matrices defined on a probability space. If $\mathbf{Y}_n$ converges in distribution to a random variable $\mathbf{Y}$ (written as $\mathbf{Y}_n \xrightarrow{d} \mathbf{Y}$), and $\mathbf{Z}_n$ converges in probability to a constant deterministic matrix $\mathbf{C}$ (written as $\mathbf{Z}_n \xrightarrow{p} \mathbf{C}$), then:

$$\mathbf{Z}_n \mathbf{Y}_n \xrightarrow{d} \mathbf{C} \mathbf{Y}$$$$\mathbf{Y}_n + \mathbf{Z}_n \xrightarrow{d} \mathbf{Y} + \mathbf{C}$$

**2. Operationalizing the Parametric Distribution via Slutsky's Theorem**

The Multivariate Central Limit Theorem ($\text{MCLT}$) establishes that:
$$\mathbf{Y}_n = \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$$

We can pre-multiply this expression by the inverse principal square root of the latent population covariance matrix, $\boldsymbol{\Sigma}^{-1/2}$, to standardize the distribution:
$$\boldsymbol{\Sigma}^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \xrightarrow{d} \mathscr{N}(\mathbf{0}, \mathbf{I}_{m \times m})$$

Since the population covariance matrix $\boldsymbol{\Sigma}$ is unobservable, we use the sample covariance matrix $\mathbf{S}$. Weak Law of Large Numbers ($\text{WLLN}$) states that the sample covariance matrix acts as a consistent estimator for the population covariance matrix, meaning it converges in probability to $\boldsymbol{\Sigma}$ as the sample size grows:
$$\mathbf{S} \xrightarrow{p} \boldsymbol{\Sigma} \quad \text{as } n \to \infty$$

Because matrix inversion and matrix square roots are continuous operations away from singularity, Slutsky's continuous mapping theorem preserves this convergence:
$$\mathbf{S}^{-1/2} \xrightarrow{p} \boldsymbol{\Sigma}^{-1/2} \quad \text{as } n \to \infty$$

Now we can set up the operationalized, sample-standardized sequence:
$$\mathbf{W}_n = \mathbf{S}^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] = \left( \mathbf{S}^{-1/2} \boldsymbol{\Sigma}^{1/2} \right) \left( \boldsymbol{\Sigma}^{-1/2} \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right)$$

Applying Slutsky's Theorem to this product:

The left matrix term converges in probability to the identity matrix: $\mathbf{S}^{-1/2}\boldsymbol{\Sigma}^{1/2} \xrightarrow{p} \mathbf{I}_{m \times m}$.

The right vector term converges in distribution to a standard multivariate normal distribution: $\boldsymbol{\Sigma}^{-1/2}\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathscr{N}(\mathbf{0}, \mathbf{I}_{m \times m})$.

Therefore, the combined sequence converges to:$$\mathbf{W}_n \xrightarrow{d} \mathbf{I}_{m \times m} \cdot \mathscr{N}(\mathbf{0}, \mathbf{I}_{m \times m}) = \mathscr{N}(\mathbf{0}, \mathbf{I}_{m \times m})$$
For large, finite sample snapshots ($n \gg 1$), we can map this asymptotic distribution back to the sample mean vector $\widehat{\boldsymbol{\mu}}_n$:$$\mathbf{S}^{-1/2} \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \sim \mathscr{N}(\mathbf{0}, \mathbf{I}_{m \times m})$$$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \sim \mathscr{N}(\mathbf{0}, \mathbf{S})$$$$\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu} \sim \mathscr{N}\left(\mathbf{0}, \frac{1}{n}\mathbf{S}\right) \implies \widehat{\boldsymbol{\mu}}_n \sim \mathscr{N}\left(\boldsymbol{\mu}, \frac{1}{n}\mathbf{S}\right)$$

This completes the proof. Slutsky's Theorem theoretically justifies substituting the empirical matrix $\mathbf{S}$ for the unobservable parameter $\boldsymbol{\Sigma}$ without losing asymptotic parametric validity.

````python
import numpy as np
import pandas as pd
import scipy.stats as stats

# --- DO NOT MODIFY: Synthetic Strain Sensor Data Generator ---
np.random.seed(42)
n_samples = 5000
n_features = 3

base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)

# Injecting a subtle, hidden structural drift in the final 1,000 snapshots
base_data[4000:, 0] += 0.015  # Sensor 1 drift
base_data[4000:, 2] -= 0.010  # Sensor 3 drift

df_strain = pd.DataFrame(base_data, columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'])
# ----------------------------------------------------------------

def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:
    """
    Partitions the dataset into g chunks and evaluates first-moment homogeneity
    via Wilks' Lambda and Bartlett's Chi-Square asymptotic transformation.
    """
    X = df.to_numpy()
    n, m = X.shape

    # 1. Split matrix into g equal consecutive blocks
    chunk_size = n // g_chunks
    chunks = [X[j * chunk_size : (j + 1) * chunk_size] for j in range(g_chunks)]

    # 2. Compute the global mean and individual chunk means
    global_mean = np.mean(X, axis=0)
    chunk_means = [np.mean(chunk, axis=0) for chunk in chunks]

    # 3. Calculate Within-Chunk Matrix (W) and Between-Chunk Matrix (B) from scratch
    W = np.zeros((m, m))
    B = np.zeros((m, m))

    for j in range(g_chunks):
        chunk = chunks[j]
        c_mean = chunk_means[j]
        n_j = chunk.shape[0]

        # Within matrix update: Sum squared deviations inside the current chunk
        dev_within = chunk - c_mean
        W += dev_within.T @ dev_within

        # Between matrix update: Squared deviation of chunk mean from global mean
        dev_between = (c_mean - global_mean).reshape(-1, 1)
        B += n_j * (dev_between @ dev_between.T)

    # 4. Compute Wilks' Lambda: |W| / |B + W|
    det_W = np.linalg.det(W)
    det_Total = np.linalg.det(B + W)
    wilks_lambda = det_W / det_Total

    # 5. Apply Bartlett's Chi-Square Transformation
    multiplier = -(n - 1 - (m + g_chunks) / 2.0)
    chi2_calc = multiplier * np.log(wilks_lambda)

    # 6. Evaluate p-value under Chi-Square distribution with m * (g - 1) DoF
    dof = m * (g_chunks - 1)
    p_value = 1.0 - stats.chi2.cdf(chi2_calc, dof)

    conclusion = (
        "REJECT H0: First moment is heterogeneous (Stationary baseline center shifted over time)."
        if p_value < 0.05 else
        "FAIL TO REJECT H0: First moment is homogeneous (Stationary baseline condition holds)."
    )

    return {
        "Wilks' Lambda (𝚲)": float(wilks_lambda),
        "Bartlett Chi-Square (𝛘²_calc)": float(chi2_calc),
        "Degrees of Freedom": int(dof),
        "p-value": float(p_value),
        "Diagnostic Conclusion": conclusion
    }

# Execute function
results = verify_first_moment_homogeneity(df_strain, g_chunks=5)
for key, val in results.items():
    print(f"{key}: {val}" if not isinstance(val, float) else f"{key}: {val:.7f}")
````

The coordinate projection mapping is defined as:$$\mathbf{Z}_i = \mathbf{P}^T \widetilde{\mathbf{X}}_i$$

To find the theoretical covariance matrix of this transformed coordinate space, we evaluate the expected outer product of $\mathbf{Z}_i$:$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}\left[ \left(\mathbf{P}^T \widetilde{\mathbf{X}}_i\right) \left(\mathbf{P}^T \widetilde{\mathbf{X}}_i\right)^T \right]$$

Applying the transpose product rule $(\mathbf{A}\mathbf{B})^T = \mathbf{B}^T\mathbf{A}^T$:$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}\left[ \mathbf{P}^T \widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T \mathbf{P} \right]$$

Since the eigenvector matrix $\mathbf{P}$ is deterministic, it can pass through the expectation operator:$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \mathbb{E}[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T] \mathbf{P}$$

Substituting the known population covariance matrix $\boldsymbol{\Sigma}$ into the equation:$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \boldsymbol{\Sigma} \mathbf{P}$$

Now, substituting the spectral decomposition $\boldsymbol{\Sigma} = \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T$ into the expression:$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \left( \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T \right) \mathbf{P}$$

Regrouping the matrix transformations using associative multiplication:$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \left(\mathbf{P}^T \mathbf{P}\right) \mathbf{\Lambda} \left(\mathbf{P}^T \mathbf{P}\right)$$

Since $\mathbf{P}$ is orthogonal, $\mathbf{P}^T \mathbf{P} = \mathbf{I}$. This simplifies the expression down to:$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{I} \mathbf{\Lambda} \mathbf{I} = \mathbf{\Lambda}$$

**Geometric and Statistical Meaning of this Result**

This diagonal result shows that the transformation completely changes the coordinate system. Statistically, it shows that the cross-covariance between different principal coordinates is exactly zero:
$$\text{Cov}(Z_{i,j}, Z_{i,k}) = 0 \quad \forall j \neq k$$
Geometrically, this means that the PCA transformation rotates the original sensor axes to line up with the principal orthogonal axes of the data's concentration ellipsoid. This completely de-correlates the features, ensuring that each axis captures a completely independent dimension of variance.

**2. Conservation of Energy (Total Variance Preservation)**

The total system variance in the original coordinate space is equal to the sum of the variances of each individual sensor channel, which matches the trace of the covariance matrix, $\text{tr}(\boldsymbol{\Sigma})$. We use the spectral decomposition and the cyclic permutation property of the trace operator ($\text{tr}(\mathbf{A}\mathbf{B}\mathbf{C}) = \text{tr}(\mathbf{C}\mathbf{A}\mathbf{B})$) to evaluate this value:
$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{P} \mathbf{\Lambda} \mathbf{P}^T)$$Permuting the matrix components inside the trace operator:$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{P}^T \mathbf{P} \mathbf{\Lambda})$$Since the eigenvector matrix is orthogonal, substituting $\mathbf{P}^T \mathbf{P} = \mathbf{I}$ gives:$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{I} \mathbf{\Lambda}) = \text{tr}(\mathbf{\Lambda}) = \sum_{j=1}^m \lambda_j$$

This mathematical identity proves that total system variance is perfectly preserved during the transformation. PCA simply redistributes the original system energy without losing any information.

**Mathematical Formulations for Subspace Ratios**

When truncating our model to $k$ components using the empirical sample eigenvalues $\widehat{\lambda}_1 \ge \widehat{\lambda}_2 \ge \dots \ge \widehat{\lambda}_m$:

Cumulative Explained Variance Ratio $\Phi(k)$: Measures the percentage of total system energy captured within the active principal subspace.$$\Phi(k) = \frac{\sum_{j=1}^k \widehat{\lambda}_j}{\sum_{g=1}^m \widehat{\lambda}_g} = \frac{\text{tr}(\widehat{\mathbf{P}}_k^T \mathbf{S} \widehat{\mathbf{P}}_k)}{\text{tr}(\mathbf{S})}$$

Residual Unexplained Variance Ratio $\Psi(k)$: Measures the percentage of total energy discarded or left behind in the orthogonal residual noise subspace.$$\Psi(k) = 1 - \Phi(k) = \frac{\sum_{j=k+1}^m \widehat{\lambda}_j}{\sum_{g=1}^m \widehat{\lambda}_g} = \frac{\text{tr}(\widehat{\mathbf{P}}_{m-k}^T \mathbf{S} \widehat{\mathbf{P}}_{m-k})}{\text{tr}(\mathbf{S})}$$

**3. Orthogonal Subspace Decompositions ($T^2$ versus $Q$)**

Given the truncated coordinate mapping system, the reconstructed vector $\widehat{\mathbf{x}}_i$ and its corresponding residual error projection vector $\mathbf{e}_i$ are defined as:$$\widehat{\mathbf{x}}_i = \widehat{\boldsymbol{\mu}}_n + \widehat{\mathbf{P}}_k \mathbf{z}_{i,k}$$$$\mathbf{e}_i = (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n) - \widehat{\mathbf{x}}_i = \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$$

**Pythagorean Geometric Decomposition Proof**

To show how the total squared Euclidean norm splits into independent parts, we write out the squared norm as a vector transpose product:$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \left( \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} \right)^T \left( \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} \right)$$

Expanding this product:$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} + \mathbf{z}_{i,m-k}^T \widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$$

By definition, the principal eigenvectors are mutually orthogonal. This gives us two key properties:

Inner subspace alignments are identities: $\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k = \mathbf{I}_{k \times k}$ and $\widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} = \mathbf{I}_{(m-k) \times (m-k)}$.

Cross subspace projections are zero: $\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} = \mathbf{0}_{k \times (m-k)}$ and $\widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_k = \mathbf{0}_{(m-k) \times k}$.

Substituting these properties back into our expanded equation cancels out the cross-product terms:$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \mathbf{z}_{i,k}^T \left(\mathbf{I}\right) \mathbf{z}_{i,k} + \mathbf{0} + \mathbf{0} + \mathbf{z}_{i,m-k}^T \left(\mathbf{I}\right) \mathbf{z}_{i,m-k}$$$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \|\mathbf{z}_{i,k}\|^2 + \|\mathbf{z}_{i,m-k}\|^2$$This completes the proof. The total energy decomposes cleanly into the energy of the principal subspace plus the energy of the residual subspace.

**Contrasting Hotelling's $T^2$ and Residual $Q$ (SPE) Statistics**

Hotelling's $T^2$: Measures the Mahalanobis distance within the principal subspace. It monitors variations along the dominant structural directions of the model.
$$T^2_i = \mathbf{z}_{i,k}^T \mathbf{\Lambda}_k^{-1} \mathbf{z}_{i,k} = \sum_{j=1}^k \frac{z_{i,j}^2}{\lambda_j}$$

Residual $Q$ (SPE): Measures the squared Euclidean distance within the residual noise subspace. It monitors variations that fall outside the baseline model space.$$Q_i = \|\mathbf{e}_i\|^2 = \mathbf{z}_{i,m-k}^T \mathbf{z}_{i,m-k} = \sum_{j=k+1}^m z_{i,j}^2$$

**Matching Event Types to Diagnostic Metrics**

Environmental Load Anomaly (e.g., Extreme Wind Loads): This event maps directly to Hotelling's $T^2$. High operational wind loads do not alter the physical composition of the structure; they simply exert greater external forces. This forces the system to respond with larger amplitudes along its established baseline principal components, causing a major spike within the primary subspace ($T^2$) while the residual energy ($Q$) remains normal.

Internal Structural Fracture (e.g., Fatigue Crack): This event maps directly to the Residual $Q$ Statistic. A physical fracture changes the structural mechanics and load pathways of the asset. Because the underlying physical system has changed, the sensor correlations break away from the baseline model. The resulting data projects directly onto the discarded residual vectors, triggering a clear spike in the $Q$ statistic.

Substituting these terms into the projection identity yields:$$\mathbf{f}_i = \mathbb{E}[\mathbf{F}_i \mid \mathbf{z}_i] = \mathbf{0} + \boldsymbol{\Lambda}^T \mathbf{R}^{-1} (\mathbf{z}_i - \mathbf{0})$$$$\mathbf{f}_i = \boldsymbol{\Lambda}^T \mathbf{R}^{-1} \mathbf{z}_i$$This completes the formal mathematical derivation of Thomson's regression estimator.

````python
import numpy as np
import pandas as pd
import plotly.subplots as sp
import plotly.graph_objects as go
from sklearn.decomposition import FactorAnalysis

# --- Synthetic Data Generation (Part B Validation) ---
np.random.seed(42)
n_samples = 2500
f1 = np.random.normal(0, 1, n_samples)
f2 = np.random.normal(0, 1, n_samples)

s1 = 0.85 * f1 + 0.10 * f2 + np.random.normal(0, 0.3, n_samples)
s2 = 0.80 * f1 + 0.15 * f2 + np.random.normal(0, 0.35, n_samples)
s3 = 0.12 * f1 + 0.90 * f2 + np.random.normal(0, 0.25, n_samples)
s4 = 0.02 * f1 + 0.05 * f2 + np.random.normal(0, 1.40, n_samples)

df_asset = pd.DataFrame(
    data=np.vstack([s1, s2, s3, s4]).T,
    columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4']
)

def run_factor_analysis_diagnostic_pipeline(df: pd.DataFrame, k_factors: int = 2):
    """
    Executes a high-fidelity Factor Analysis pipeline with Varimax rotation,
    applies data processing guardrails, and renders a 2x2 Plotly subplot dashboard.
    """
    # -------------------------------------------------------------------------
    # 1. DEVELOP MATHEMATICAL ENGINE & PRE-PROCESSING GUARDRAILS
    # -------------------------------------------------------------------------
    numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    X_raw = df[numeric_cols].to_numpy()
    n, m = X_raw.shape

    # Validation Rules
    if n <= m:
        raise ValueError(f"Operational constraint violated: Snapshot count n ({n}) must exceed channel count m ({m}).")
    if k_factors >= m:
        raise ValueError(f"Dimensionality exception: Target factors k ({k_factors}) must be strictly less than features m ({m}).")

    # Z-Score Mapping Layer with Zero-Variance Caps
    means = np.mean(X_raw, axis=0)
    stds = np.std(X_raw, axis=0, ddof=1)
    stds = np.where(stds == 0, 1e-15, stds)  # Guardrail cap

    Z = (X_raw - means) / stds

    # -------------------------------------------------------------------------
    # 2. INFORMATION DECOMPOSITION LAYER (FA Engine + Varimax)
    # -------------------------------------------------------------------------
    # Fit scikit-learn FactorAnalysis model (uses Maximum Likelihood)
    fa = FactorAnalysis(n_components=k_factors, rotation='varimax', random_state=42)
    fa.fit(Z)

    # Extracted Rotated Loadings Matrix (m x k)
    Lambda_rotated = fa.components_.T

    # Compute Empirical Variance-Covariance Array R
    R_empirical = np.corrcoef(Z.T)

    # Communalities (h²) and Uniquenesses (𝜪²) Vectors
    communality = np.sum(Lambda_rotated**2, axis=1)
    uniqueness = np.clip(1.0 - communality, 0.0, 1.0)

    # -------------------------------------------------------------------------
    # 3. LATENT SCORE PROJECTION (Thomson's MMSE Regression Method)
    # -------------------------------------------------------------------------
    R_inv = np.linalg.inv(R_empirical)
    # Thomson Scoring Weights Matrix: R^-1 @ Lambda
    W_thomson = R_inv @ Lambda_rotated
    F_scores = Z @ W_thomson

    # Calculate Empirical Variances of Factor Scores (ddof=1)
    factor_variances = np.var(F_scores, axis=0, ddof=1)

    # Performance Diagnostic Telemetry
    avg_communality = np.mean(communality) * 100
    avg_uniqueness = np.mean(uniqueness) * 100
    print(f"--- PERFORMANCE DIAGNOSTIC TELEMETRY ---")
    print(f"Average System Communality %: {avg_communality:.3f}%")
    print(f"Average System Uniqueness % : {avg_uniqueness:.3f}%\n")

    # -------------------------------------------------------------------------
    # 4. LAYOUT ENGINEERING & INTERFACE DESIGN (2x2 Plotly Canvas)
    # -------------------------------------------------------------------------
    fig = sp.make_subplots(
        rows=2, cols=2,
        horizontal_spacing=0.24,
        vertical_spacing=0.28,
        subplot_titles=(
            "<b>Structural Loadings Matrix Heatmap</b>",
            "<b>Variance Partitioning Profile</b>",
            "<b>Sensor Uniqueness Line Profile</b>",
            "<b>Latent Factor Variance Distribution</b>"
        )
    )

    # Subplot (1,1): Structural Loadings Heatmap (|𝚲_rotated|)
    abs_loadings = np.abs(Lambda_rotated)
    factor_labels = [f"Factor {j+1}" for j in range(k_factors)]

    heatmap = go.Heatmap(
        z=abs_loadings,
        x=factor_labels,
        y=numeric_cols,
        colorscale='YlOrRd',
        colorbar=dict(
            x=-0.15, len=0.38, y=0.78,
            yanchor="middle", xanchor="right",
            title="Abs Weight"
        ),
        showscale=True
    )
    fig.add_trace(heatmap, row=1, col=1)

    # Subplot (1,2): Variance Partitioning Profile (Horizontal Stacked Bar)
    fig.add_trace(go.Bar(
        y=numeric_cols, x=communality * 100,
        name="Shared Communality (h²)",
        orientation='h', marker=dict(color='#1f77b4')
    ), row=1, col=2)

    fig.add_trace(go.Bar(
        y=numeric_cols, x=uniqueness * 100,
        name="Isolated Uniqueness (𝜪²)",
        orientation='h', marker=dict(color='#ff7f0e')
    ), row=1, col=2)

    # Subplot (2,1): Sensor Uniqueness Line Profile
    fig.add_trace(go.Scatter(
        x=numeric_cols, y=uniqueness,
        mode='lines+markers',
        name="Uniqueness (𝜪²)",
        line=dict(color='#d62728', width=2, dash='dashdot'),
        marker=dict(symbol='x', size=8)
    ), row=2, col=1)

    # Subplot (2,2): Latent Factor Variance Distribution
    fig.add_trace(go.Bar(
        x=factor_labels, y=factor_variances,
        name="Factor Variance",
        marker=dict(color='#2ca02c', line=dict(color='black', width=0.5))
    ), row=2, col=2)

    # Apply Global Canvas Layout Constraints
    fig.update_layout(
        title=dict(
            text="<b>PRODUCTION-GRADE FA LATENT DIAGNOSTICS SUITE</b>",
            x=0.5, y=0.97, font=dict(size=16)
        ),
        template="plotly_white",
        width=1250, height=750,
        margin=dict(t=150, b=60, l=140, r=80),
        barmode='stack',
        legend=dict(
            orientation="h",
            x=0.5, y=1.04,
            xanchor="center", yanchor="bottom"
        )
    )

    # Apply axis updates and constraints
    fig.update_xaxes(title_text="Latent Cutoff", row=1, col=1)
    fig.update_xaxes(title_text="Variance Allocation %", range=[0, 100], row=1, col=2)
    fig.update_xaxes(title_text="Monitored Channels", tickangle=25, row=2, col=1)
    fig.update_yaxes(title_text="Uniqueness Metric Value", row=2, col=1)
    fig.update_xaxes(title_text="Extracted Common Factors", row=2, col=2)
    fig.update_yaxes(title_text="Empirical Variance Score", row=2, col=2)

    fig.show()

# Run optimization framework
run_factor_analysis_diagnostic_pipeline(df_asset, k_factors=2)
````

````python
import numpy as np
import pandas as pd
import plotly.subplots as sp
import plotly.graph_objects as go
from sklearn.decomposition import FactorAnalysis

# --- RUNNING DUAL FEATURE ANALYSIS INTERFACE SUITE ---
def generate_dual_feature_analysis_dashboards(df: pd.DataFrame):
    X_raw = df.to_numpy()
    n, m = X_raw.shape
    Z = (X_raw - np.mean(X_raw, axis=0)) / np.std(X_raw, axis=0, ddof=1)

    # Compute Eigen Decompositions for PCA
    cov_Z = np.cov(Z.T)
    eigenvalues, eigenvectors = np.linalg.eigh(cov_Z)

    # Sort in descending order
    idx = np.argsort(eigenvalues)[::-1]
    lambdas = eigenvalues[idx]
    P_matrix = eigenvectors[:, idx]

    # Pre-calculate PCA profiles across all possible k limits
    marginal_variance = (lambdas / np.sum(lambdas)) * 100
    cumulative_variance = np.cumsum(marginal_variance)
    residual_variance = 100.0 - cumulative_variance

    mean_T2_profile = []
    mean_Q_profile = []
    k_range = [1, 2, 3]

    for k in k_range:
        # T² metrics: Sum_j=1..k (z_ij^2 / lambda_j)
        Z_pca = Z @ P_matrix[:, :k]
        T2_samples = np.sum((Z_pca**2) / lambdas[:k], axis=1)
        mean_T2_profile.append(np.mean(T2_samples))

        # Q metrics (SPE): Norm of residual error vector
        P_residual = P_matrix[:, k:]
        E_residuals = Z @ P_residual @ P_residual.T
        Q_samples = np.sum(E_residuals**2, axis=1)
        mean_Q_profile.append(np.mean(Q_samples))

    # =========================================================================
    # INTERFACE 1: PCA OPTIMIZATION SUITE (2 x 3 Layout Matrix)
    # =========================================================================
    pca_fig = sp.make_subplots(
        rows=2, cols=3,
        subplot_titles=(
            "<b>(1,1) Feature Loadings |P_jr|</b>", "<b>(1,2) Variance Eigenvalues (λ)</b>",
            "<b>(1,3) Marginal + Cumulative Curve</b>", "<b>(2,1) Residual Unexplained %</b>",
            "<b>(2,2) Mean Hotelling T² vs k</b>", "<b>(2,3) Mean Q Statistic vs k</b>"
        ),
        horizontal_spacing=0.10, vertical_spacing=0.25
    )

    pc_labels = [f"PC {j+1}" for j in range(m)]
    sensor_labels = df.columns.tolist()

    pca_fig.add_trace(go.Heatmap(z=np.abs(P_matrix), x=pc_labels, y=sensor_labels, colorscale='Viridis', showscale=False), row=1, col=1)
    pca_fig.add_trace(go.Bar(x=pc_labels, y=lambdas, marker=dict(color='#2ca02c')), row=1, col=2)

    pca_fig.add_trace(go.Bar(x=pc_labels, y=marginal_variance, name="Marginal Vol %", marker=dict(color='#1f77b4')), row=1, col=3)
    pca_fig.add_trace(go.Scatter(x=pc_labels, y=cumulative_variance, name="Cumulative %", mode='lines+markers', line=dict(color='red', dash='dash')), row=1, col=3)

    k_labels = [f"k={k}" for k in k_range]
    pca_fig.add_trace(go.Bar(x=[f"k={j}" for j in [1, 2, 3, 4]], y=np.append(residual_variance, 0), marker=dict(color='#9467bd')), row=2, col=1)
    pca_fig.add_trace(go.Scatter(x=k_labels, y=mean_T2_profile, mode='lines+markers', line=dict(color='black', width=2)), row=2, col=2)
    pca_fig.add_trace(go.Scatter(x=k_labels, y=mean_Q_profile, mode='lines+markers', line=dict(color='orange', width=2)), row=2, col=3)

    pca_fig.update_layout(title="<b>INTERFACE SUITE 1: PCA GEOMETRIC SYSTEM OPTIMIZATION ARCHITECTURE</b>", template="plotly_white", width=1300, height=700)
    pca_fig.show()

generate_dual_feature_analysis_dashboards(df_asset)
````

### Solution for Question 1: The Total Variance Illusion

**Physical Nature of Sensor_4**: A Uniqueness value ($\varphi^2$) of over $98\%$ indicates that Sensor_4 has decoupled from the shared physical system. The data it produces contains almost no information about the underlying asset dynamics ($f_1$ and $f_2$). Instead, it is generating independent, localized white noise, confirming a severe instrumentation or electrical fault.

**How the PCA Engine is "Tricked":** PCA calculates its principal directions by maximizing total global variance, without distinguishing between useful shared variance and random localized noise. Because Sensor_4 has a massive noise scale, its variance ($\sigma^2 \approx 2.0$) is the largest single source of energy in the system. To capture this energy, the PCA engine aligns its top components with the noise from Sensor_4, inflating the first two eigenvalues (PC 1 and PC 2).

**Automated Plant Monitoring Risks:**
Relying only on PCA creates a major blind spot. An operator looking at the high explained variance of PC 1 and PC 2 might assume the model is tracking the plant's health. In reality, the automated system is monitoring sensor noise. If a true structural failure occurs, it may go unnoticed because the diagnostic layer's energy is dominated by the faulty instrument.

### Solution for Question 2: Decoupling Structural Loading via Rotation

**Varimax Rotation Mechanics:**
Traditional PCA applies a strict hierarchical constraint ($\lambda_1 > \lambda_2 > \dots$), forcing early components to absorb mixed variations from all available sensors. This leads to non-sparse eigenvectors where every channel has partial weights. Factor Analysis with Varimax rotation relaxes this hierarchical constraint. It rotates the factor space to maximize the variance of the squared loadings within each column. This drives the weights toward either high values or near-zero values, creating a "simple structure" that separates the sensor contributions into clean, distinct groups.

**Troubleshooting from a Plant Operator's Standpoint:** The rotated FA heatmap is much easier to interpret during a failure. Instead of showing a complex mix of weights across all channels, the Varimax rotation clearly isolates the subsystems: Sensor_1 and Sensor_2 map exclusively to Factor 1 (the primary structural load), while Sensor_3 maps to Factor 2 (the operational mode). If Factor 1 triggers an alarm, the operator instantly knows the issue is localized to the primary structural load pathway, allowing them to troubleshoot the problem immediately.

### Solution for Question 3: Determining Subspace Truncation ($k$)

**Behavior of the Mean $Q$ Statistic Curve:** As the truncation boundary moves from $k=1$ to $k=2$, the Mean $Q$ statistic drops sharply. This indicates that adding the second component captures a large amount of systematic variance. However, when moving from $k=2$ to $k=3$, the curve flattens completely into a distinct "elbow," meaning that adding a third component provides almost no additional reduction in residual energy.

**Identifying the True System Dimensionality ($k=2$):** The sharp drop followed by a flat elbow at $k=2$ marks the transition from systematic structural variance to random noise. The first two dimensions capture the actual physical driving forces ($f_1$ and $f_2$). Once these are accounted for, the remaining variation is purely random noise, confirming that $k=2$ is the true underlying dimension of the asset.

**Consequences of Choosing an Over-Extended Subspace ($k=3$):** If an engineer sets the model truncation to $k=3$, they force the unmodeled residual noise into the "clean" principal subspace. This overfits the model, corrupting the $T^2$ statistic with random instrumentation errors and making the monitoring system unstable and prone to false alarms.

### Solution for Question 4: Operational Trade-offs in System Health Monitoring

**Sensitivity to Local Instrument Faults**

The two strategies handle a localized sensor failure—such as a sudden calibration drift or an electrical short—through entirely different mathematical channels.

The PCA Strategy: When a sensor malfunctions, the sudden burst of localized energy immediately leaks into the residual subspace. This triggers a sharp, immediate spike in the residual $Q$ monitor (also known as the Squared Prediction Error), alerting operators that the data no longer fits the baseline model.

The FA Strategy: Instead of throwing an alarm across the entire system, the Factor Analysis engine absorbs the localized error directly into its unique noise matrix ($\mathbf{\Psi}$). It recognizes that the anomaly is isolated to a single channel rather than a shared physical shift, keeping the core tracking metrics steady.

**Robustness Against Signal Corruption**

How well each strategy protects the overall diagnostic system from being compromised by a single faulty asset sensor.

The PCA Strategy: This approach offers mixed robustness. Because PCA builds its geometric coordinate system by maximizing total global variance, a sensor with massive localized noise can easily warp the principal vectors. This distortion shifts the baseline asset model, potentially causing false alarms or masking real failures in other healthy channels.

The FA Strategy: This approach offers high robustness. Because the generative model separates shared variance from unique variance, it essentially filters out localized sensor noise before calculating any factor scores. This isolation ensures that a single broken instrument cannot corrupt the latent scores tracking the rest of the healthy plant.

**Physical Interpretability**

The ease with which an operations team can trace a statistical alarm back to a specific physical root cause on the plant floor.

The PCA Strategy: This approach has low interpretability. Because PCA forces sensors into a strict mathematical hierarchy, the resulting principal components are dense, complex combinations of almost all available variables. When an alarm triggers, engineers must spend critical time manually deconstructing the eigenvectors to figure out which subsystem is actually failing.

The FA Strategy: This approach has high interpretability. By utilizing orthogonal rotations like Varimax, the model achieves a "simple structure" where specific sensor groups map directly to distinct, independent physical factors (such as isolating structural load from operational speed). When a specific factor spikes, operators can instantly point to the exact physical subsystem that requires maintenance.
